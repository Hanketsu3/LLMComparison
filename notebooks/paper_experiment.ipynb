```python
import json
from pathlib import Path
import pandas as pd
rows=[]
for p in Path('results').glob('*/aggregate_metrics.json'):
    payload=json.loads(p.read_text())
    for model_name, metrics in payload.get('metrics', {}).items():
        rows.append({'run_name': p.parent.name, 'model_name': model_name, **metrics})
df=pd.DataFrame(rows)
metric_cols=[c for c in df.columns if c.endswith('_mean')]
display(df.sort_values(metric_cols[0], ascending=False) if (not df.empty and metric_cols) else df)
```
