# Colab Setup
Bu notebook projeyi clone eder, bagimliliklari kurar ve deney notebooklarina baz olacak yolu hazirlar.

```python
import os
import subprocess
import sys

REPO_URL = "https://github.com/Hanketsu3/LLMComparison.git"
BRANCH = "main"
PROJECT_DIR = '/content/LLMComparison'
INSTALL_REQUIREMENTS = True
INSTALL_PACKAGE = True

print(f"Repo URL: {REPO_URL}")
print(f"Project dir: {PROJECT_DIR}")
```

```python
from pathlib import Path

PROJECT_DIR = Path(PROJECT_DIR)

def clone_repo() -> None:
    clone_commands = [
        ["git", "clone", "--depth", "1", "--single-branch", "--branch", BRANCH, REPO_URL, str(PROJECT_DIR)],
        ["git", "clone", "--depth", "1", "--single-branch", REPO_URL, str(PROJECT_DIR)],
    ]

    last_error = None
    for command in clone_commands:
        print('Trying clone command:', ' '.join(command))
        try:
            subprocess.run(command, check=True)
            return
        except subprocess.CalledProcessError as exc:
            last_error = exc
            print('Clone attempt failed, trying fallback if available.')
            print(exc)

    raise last_error

if PROJECT_DIR.exists():
    if (PROJECT_DIR / '.git').exists():
        print('Repo already exists, skipping clone.')
    else:
        raise FileExistsError(
            f'{PROJECT_DIR} exists but is not a git repo. Delete it manually and rerun this cell.'
        )
else:
    clone_repo()

os.chdir(PROJECT_DIR)
print(f"Working directory: {Path.cwd()}")
```

```python
if INSTALL_REQUIREMENTS:
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)

if INSTALL_PACKAGE:
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-e', '.'], check=True)

print('Dependencies installed.')
```
