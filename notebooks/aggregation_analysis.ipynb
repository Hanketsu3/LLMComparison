# Aggregation Analysis
Saved outputs uzerinden tablo, heatmap ve istatistik incelemesi.

```python
import json
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
rows = []
for p in Path('results').glob('*/aggregate_metrics.json'):
    payload = json.loads(p.read_text())
    for model_name, metrics in payload.get('metrics', {}).items():
        rows.append({'run_name': p.parent.name, 'model_name': model_name, **metrics})
df = pd.DataFrame(rows)
display(df.head())
```

```python
if not df.empty:
    metric_cols = [c for c in df.columns if c.endswith('_mean')]
    if metric_cols:
        pivot = df.pivot_table(index='model_name', values=metric_cols[0], aggfunc='mean')
        sns.heatmap(pivot, annot=True, fmt='.3f', cmap='viridis')
        plt.title(metric_cols[0])
        plt.show()
```
