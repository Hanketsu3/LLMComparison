# 99 Result Aggregation
Model notebooklarindan gelen standart JSON dosyalarini birlestirir.

```python
import json
from pathlib import Path
import pandas as pd
rows = []
for aggregate_file in Path('results').glob('*/aggregate_metrics.json'):
    run_name = aggregate_file.parent.name
    payload = json.loads(aggregate_file.read_text())
    for model_name, metrics in payload.get('metrics', {}).items():
        rows.append({'run_name': run_name, 'model_name': model_name, **metrics})
df = pd.DataFrame(rows)
if df.empty:
    print('No aggregate metrics found yet.')
else:
    df = df.sort_values(['run_name', 'model_name']).reset_index(drop=True)
    display(df)
    metric_columns = [column for column in df.columns if column not in {'run_name', 'model_name'}]
    print('Rows:', len(df))
    print('Metric columns:', ', '.join(metric_columns))
```
